# TASK 02 – DATASET UNDERSTANDING

### Step 1 – Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("Libraries imported successfully.")
```

### Step 2 – Load Dataset

```python
df = pd.read_csv("smartcare_ai_dataset_1000.csv")

print("Dataset loaded successfully.")
```

### Step 3 – Display First Five Records

```python
df.head()
```

### Step 4 – Display Last Five Records

```python
df.tail()
```

### Step 5 – Dataset Dimensions

```python
rows, columns = df.shape

print("Number of Records:", rows)
print("Number of Attributes:", columns)
print("Dataset Shape:", df.shape)
```

### Step 6 – List All Attributes

```python
print("Total number of attributes:", len(df.columns))
print("\nAttribute Names:\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i:02d}. {column}")
```

### Step 7 – Dataset Information and Data Types

```python
df.info()
```

### Step 8 – Data Type Summary

```python
dtype_summary = df.dtypes.value_counts()

print("Data Type Summary:")
print(dtype_summary)
```

### Step 9 – Numerical Attributes

```python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

print("Number of numerical attributes:", len(numerical_columns))
print("\nNumerical Attributes:")

for column in numerical_columns:
    print("-", column)
```

### tep 10 – Categorical Attributes

```python
categorical_columns = df.select_dtypes(include=['object', 'category']).columns.tolist()

print("Number of categorical attributes:", len(categorical_columns))
print("\nCategorical Attributes:")

for column in categorical_columns:
    print("-", column)
```

### Step 11 – Statistical Summary

```python
df.describe().T
```

### Step 12 – Categorical Attribute Summary

```python
for column in categorical_columns:
    print("=" * 60)
    print(f"Attribute: {column}")
    print("Number of unique values:", df[column].nunique(dropna=True))
    print("\nValue counts:")
    print(df[column].value_counts(dropna=False))
    print()
```

### Step 13 – Identify Input and Target Variables

```python
# Define target variable
target_column = "no_show"

# Identify input variables
input_columns = [column for column in df.columns if column != target_column]

print("Target Variable:")
print(target_column)

print("\nNumber of Input Variables:", len(input_columns))

print("\nInput Variables:")
for column in input_columns:
    print("-", column)
```

### Step 14 – Verify Target Variable

```python
print("Target Variable:", target_column)
print("Data Type:", df[target_column].dtype)
print("Number of Unique Values:", df[target_column].nunique())

print("\nUnique Values:")
print(df[target_column].unique())

print("\nTarget Value Counts:")
print(df[target_column].value_counts(dropna=False))
```

### Step 15 – Target Class Distribution

```python
target_counts = df[target_column].value_counts().sort_index()

print("Target Class Distribution:")
print(target_counts)

print("\nTarget Class Percentages:")
print((target_counts / len(df) * 100).round(2))
```

### Step 16 – Target Distribution Visualization

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x=target_column
)

plt.title("Distribution of Appointment No-Show")
plt.xlabel("No Show")
plt.ylabel("Number of Patients")

# Add values on top of bars
for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

### Step 17 – Missing Value Analysis

```python
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage.round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

missing_summary
```

### Step 19 – Duplicate Record Analysis

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

### Step 20 – Display Duplicate Records

```python
duplicates = df[df.duplicated(keep=False)]

print("Duplicate records found:", len(duplicates))

duplicates.head(20)
```

### Step 21 – Unique Values / Cardinality

```python
unique_summary = pd.DataFrame({
    "Attribute": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Unique Values": [df[column].nunique(dropna=True) for column in df.columns]
})

unique_summary
```

### Step 23 – Check Missing Target Values

```python
print("Missing target values:", df[target_column].isnull().sum())

print("\nTarget values:")
print(df[target_column].value_counts(dropna=False))
```

### Step 24 – Final Dataset Overview

```python
print("=" * 70)
print("TASK 02 – DATASET UNDERSTANDING SUMMARY")
print("=" * 70)

total_missing = df.isnull().sum().sum()
total_duplicates = df.duplicated().sum()
missing_target = df[target_column].isnull().sum()

print(f"Dataset Size       : {df.shape}")
print(f"Total Records      : {len(df)}")
print(f"Total Attributes   : {len(df.columns)}")
print(f"Numerical Features : {len(numerical_columns)}")
print(f"Categorical Features: {len(categorical_columns)}")
print(f"Missing Values     : {total_missing}")
print(f"Duplicate Records  : {total_duplicates}")
print(f"Target Variable    : {target_column}")
print(f"Missing Target     : {missing_target}")

print("\nTask 02 completed successfully.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

ax = sns.countplot(x='no_show', data=df)

total = len(df)

for p in ax.patches:
    percentage = p.get_height() / total * 100
    ax.annotate(f'{percentage:.1f}%',
                (p.get_x() + p.get_width() / 2, p.get_height()),
                ha='center', va='bottom')

plt.title('Appointment No-Show Distribution')
plt.xlabel('No-Show (0 = Attended, 1 = No-Show)')
plt.ylabel('Number of Patients')
plt.show()
```
